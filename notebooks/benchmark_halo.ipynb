## Visual inspection of benchmark halo snaps

```python
def plot_benchmark_halo(halo_array, cell_coms):
    """
    Plot dark matter halos in 3D.
    
    Parameters:
    halo_array (numpy.ndarray): Array of shape (N, 3) containing x, y, z coordinates of halos.
    """
    # Create a new 3D figure
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Extract x, y, z coordinates
    x, y, z = halo_array.T
    
    # Create the 3D scatter plot
    ax.scatter(x, y, z, c='purple', s=0.01, alpha=0.2, label="DM particles")
    
    # Extract x, y, z coordinates for cell centers
    cx, cy, cz = cell_coms.T/Params.kpc
    
    # Add cell centers as black crosses
    ax.scatter(cx, cy, cz, c='black', marker='x', s=5, alpha=0.4, label='cell c.o.m.')

    # Plot styling
    ax.set_xlabel('X [kpc]')
    ax.set_ylabel('Y [kpc]')
    ax.set_zlabel('Z [kpc]')
    ax.set_title('Benchmark (NFW) halo')
    ax.legend()
    
    # Show the plot
    plt.show()


snap = jnp.load(f"data/benchmark_halo_files/benchmark_halo_snap_0012.npy")
print(len(snap))
coms = jnp.load(f"data/benchmark_halo_files/cell_com_benchmark_halo_snap_0012.npy")
plot_benchmark_halo(snap, coms)
snap = jnp.load(f"data/benchmark_halo_files/benchmark_halo_snap_0036.npy")
print(len(snap))
coms = jnp.load(f"data/benchmark_halo_files/cell_com_benchmark_halo_snap_0036.npy")
plot_benchmark_halo(snap, coms)
```
