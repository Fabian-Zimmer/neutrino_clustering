```python
def distances_histogram_log_bins(
    distances, Fermi_Dirac, CMB_LSS, ax_lims, fig_dir, args):
    """
    Plot a histogram of distances with logarithmically spaced bins.
    """

    # Avoiding zero distance for log scale
    distances_above0 = distances[distances > 0]

    # From kpc to Mpc
    distances_Mpc = distances_above0/1e3

    # Generating logarithmically spaced bins
    log_bins = jnp.logspace(
        jnp.log10(jnp.min(distances_Mpc)), 
        jnp.log10(jnp.max(distances_Mpc)), 
        100)
    
    fig = plt.figure()
    ax = fig.add_subplot(111)

    # Plot simulation distances
    ax.hist(
        distances_Mpc, bins=log_bins, 
        color='blue', label='CNB simulation', alpha=0.8)

    # Plot Fermi-Dirac sampled momentum comoving distances
    if Fermi_Dirac is not None:
        Fermi_Dirac_Mpc = Fermi_Dirac/1e3
        ax.hist(
            Fermi_Dirac_Mpc, bins=log_bins, 
            color='orange', label='Fermi-Dirac no-gravity', alpha=0.7)

    # Plot Distance to CMB LSS
    if CMB_LSS:
        z_end_sim = 4.0
        CMB_LSS_dist = Cosmo.get_LSS_distance(
            z_end_sim, 1.0, 0.0, Params())/(args.Mpc/args.h)
        ax.axvline(CMB_LSS_dist, c='red', ls='dashed')

        # Compute geometric mean for y-position of text
        y_lower, y_upper = ax.get_ylim()
        # y_mid = np.sqrt(y_lower * y_upper)
        y_mid = (y_upper - y_lower)/2

        # Add rotated text
        ax.text(
            CMB_LSS_dist, y_mid, 'CMB LSS', color='red', 
            rotation=90, va='center', ha='right')

    ax.set_xscale('log')
    ax.set_title("Distances at z=4 (Log Bins)")
    ax.set_xlabel("Distance [Mpc/h]")
    ax.set_ylabel("Number of Particles")
    ax.legend(loc='upper left')

    if ax_lims is not None:
        ax.set_xlim(ax_lims[0], ax_lims[1])
        ax.set_ylim(ax_lims[2], ax_lims[3])

    plt.savefig(f'{fig_dir}/distances_histogram_log_bins.pdf', bbox_inches='tight')
    plt.show(); plt.close()


# Compute CDF over a fine grid
@jax.jit
def CDF_Fermi_Dirac_grid(T, p_min, p_max):
    p_vals = jnp.linspace(p_min, p_max, 10_000)
    CDF_vals = p_vals - T*jnp.log(jnp.exp(p_vals/T)+1) + T*jnp.log(2)
    norm_factor = CDF_vals[-1]
    CDF_vals /= norm_factor
    return p_vals, CDF_vals


# Sample from the Fermi-Dirac distribution using inverse sampling
@jax.jit
def Fermi_Dirac_sampling(T, p_min, p_max, args):
    
    # Precompute CDF values
    p_vals, CDF_vals = CDF_Fermi_Dirac_grid(T, p_min, p_max)

    # Generate uniform random samples
    u_samples = jax.random.uniform(args.key, shape=(neutrinos_current,))

    # Inverse sampling using searchsorted
    indices = jnp.searchsorted(CDF_vals[:-1], u_samples)
    p_samples = p_vals[indices]
    
    return p_samples


def UTIL_select_neutrinos_momentum_filter(vectors, nu_mass, args):
    
    # Extract velocities and attach units
    zBack_velocities = vectors[..., -1, 3:6]*args.kpc/args.s

    # note on selection
    # For z0 vectors, roughly selects 65%
    # For z4 vectors, roughly selects 82%

    # Convert velocities to momentum magnitudes
    momentum = zBack_velocities * nu_mass
    mags = jnp.linalg.norm(momentum, axis=-1)

    # Filter based on momentum values
    selection = (mags >= 0.01*args.T_CNB) & (mags <= 10*args.T_CNB)
    neutrinos = jnp.count_nonzero(selection)

    percent = neutrinos/len(vectors)*100
    print(f'{neutrinos} ({percent:.1f}%) neutrinos selected.')

    return vectors[selection]


# Load sim output
halo1_vecs = jnp.load(f"{sim_folder}/vectors_halo1.npy").reshape(-1, 2, 6)
# print(halo1_vecs.shape)

# Specified neutrino selection
nu_mass_choice = 0.05*Params.eV
cut_vectors = UTIL_select_neutrinos_momentum_filter(
    halo1_vecs, nu_mass_choice, Params())
# print(cut_vectors.shape)

# Get distance magnitudes
sim_distances = jnp.linalg.norm(cut_vectors[..., -1, :3], axis=-1)
# print(sim_distances.shape)


### ------------------------------------------------------------- ###
### Generate Fermi-Dirac sampled momenta and convert to distances ###
### ------------------------------------------------------------- ###
z_end_sim = 4.0
# p_min, p_max = simdata.p_start*Params.T_CNB, simdata.p_stop*Params.T_CNB
p_min, p_max = 0.01*Params.T_CNB, 10*Params.T_CNB
neutrinos_current = len(sim_distances)

Fermi_Dirac_p_samples = Fermi_Dirac_sampling(
    Params.T_CNB, p_min, p_max, Params())
# print(Fermi_Dirac_p_samples.shape)

# Fermi_Dirac_distances = jnp.array([
#     Cosmo.get_LSS_distance(
#         z_end_sim, p_0=p, m_nu=nu_mass_choice, args=Params())
#     for p in Fermi_Dirac_p_samples])

Fermi_Dirac_distances = Cosmo.get_LSS_distance(
    z_end_sim, Fermi_Dirac_p_samples, nu_mass_choice, Params())
Fermi_Dirac_distances /= (Params.kpc/Params.h)
# print(Fermi_Dirac_distances.shape)


distances_histogram_log_bins(
    distances=sim_distances, Fermi_Dirac=Fermi_Dirac_distances, 
    CMB_LSS=True, ax_lims=None, fig_dir=fig_folder, args=Params())

# distances=Zimmer_distances_zBack, Fermi_Dirac=Fermi_Dirac_distances, 
# CMB_LSS=True)
```

### Velocity distributions

```python
def plot_cnb_velocity_distribution(
        masses, 
        unit_type: str = "km/s",  # km/s or v/c 
        args = None):
    
    if args is None:
        args = Params()

    v_unit = args.km/args.s

    # Fermi-Dirac distribution (μ = 0)
    def fermi_dirac(p):
        return 1 / (jnp.exp(p / (args.T_CNB)) + 1)
    
    # Create momentum array
    p = jnp.geomspace(0.01, 400, 10_000)*args.T_CNB
    
    # Create 2x2 subplot
    fig, axs = plt.subplots(2, 2, figsize=(12, 12))
    axs = axs.ravel()
    
    for idx, m in enumerate(masses):

        # Distribution calculation
        f = fermi_dirac(p)
        dist = f * p**3
        
        # Calculate velocities (relativistic)
        E = jnp.sqrt(p**2 + m**2)
        v = (p / E)

        if unit_type == "km/s":
            v /= v_unit
        
        # Normalize distribution
        norm = trap(dist, jnp.log(p))
        
        avg_v = trap(v*dist, jnp.log(v)) / norm
        std_v = jnp.sqrt(trap((v-avg_v)**2 * dist, jnp.log(v)) / norm)
        
        # Plot
        axs[idx].plot(v, dist/norm)
        axs[idx].set_xscale('log')

        if unit_type == "v/c":
            axs[idx].set_xlabel('v/c')
            label = f'Mean: {avg_v:.3f}c'
            axs[idx].set_xlim(1e-4, 1)
        if unit_type == "km/s":
            axs[idx].set_xlabel('km/s')
            label = f'Mean: {avg_v:.0f} km/s'
            axs[idx].set_xlim(1e1, 2e5)

        axs[idx].set_ylabel('Normalized Distribution')
        axs[idx].set_title(f'm = {m} eV')
        
        # Add mean line and 1 sigma region
        axs[idx].axvline(avg_v, color='black', linestyle='--', label=label)
        axs[idx].axvspan((avg_v-std_v), (avg_v+std_v), 
                        alpha=0.2, color='black', label=r'1$\sigma$ range')
        axs[idx].legend()

    
    plt.tight_layout()
    plt.show()


plot_cnb_velocity_distribution(
    masses=nu_m_picks[:4], 
    unit_type="km/s",
    args=Params())
```

```python
def plot_cnb_velocity_distribution_boost(
        masses, 
        unit_type: str = "km/s",
        boost_vec = jnp.array([0., 0., 0.]),
        args = None):
    
    if args is None:
        args = Params()

    v_unit = args.km/args.s
    
    def fermi_dirac(p):
        return 1 / (jnp.exp(p / (args.T_CNB)) + 1)
    
    p = jnp.geomspace(0.01, 400, 10_000)*args.T_CNB
    
    theta = jnp.arccos(2 * np.random.uniform(0, 1, size=len(p)) - 1)
    phi = jnp.array(np.random.uniform(0, 2*np.pi, size=len(p)))
    
    fig, axs = plt.subplots(2, 2, figsize=(12, 12))
    axs = axs.ravel()

    for idx, m in enumerate(masses):
        E = jnp.sqrt(p**2 + m**2)
        v_mag = p / E
        
        v = jnp.stack([
            v_mag * jnp.sin(theta) * jnp.cos(phi),
            v_mag * jnp.sin(theta) * jnp.sin(phi),
            v_mag * jnp.cos(theta)
        ], axis=1)
        
        v_boosted = SimUtil.S_to_Sprime_frame_trafo(v, boost_vec, args)
        v_boosted_mag = jnp.sqrt(jnp.sum(v_boosted**2, axis=1))
        
        sort_idx = jnp.argsort(v_boosted_mag)
        v_boosted_mag = v_boosted_mag[sort_idx]
        
        p_boosted = m * v_boosted_mag / jnp.sqrt(1 - v_boosted_mag**2)
        f = fermi_dirac(p_boosted)
        dist = f * p_boosted**3
        
        # Calculate statistics before unit conversion
        norm = trap(dist, jnp.log(p_boosted))  # integrate over log momentum
        dist_normalized = dist/norm
        
        mean_v = jnp.sum(v_boosted_mag * dist_normalized) / jnp.sum(dist_normalized)
        std_v = jnp.sqrt(jnp.sum((v_boosted_mag - mean_v)**2 * dist_normalized) / jnp.sum(dist_normalized))
        
        if unit_type == "km/s":
            v_plot = v_boosted_mag/v_unit
            mean_v_plot = mean_v/v_unit
            std_v_plot = std_v/v_unit
        else:
            v_plot = v_boosted_mag
            mean_v_plot = mean_v
            std_v_plot = std_v
            
        axs[idx].plot(v_plot, dist_normalized)
        axs[idx].set_xscale('log')

        if unit_type == "v/c":
            axs[idx].set_xlabel('v/c')
            label = f'Mean: {mean_v_plot:.3f}c'
            axs[idx].set_xlim(1e-4, 1)
        if unit_type == "km/s":
            axs[idx].set_xlabel('km/s')
            label = f'Mean: {mean_v_plot:.0f} km/s'
            axs[idx].set_xlim(1e1, 2e5)

        axs[idx].set_ylabel('Normalized Distribution')
        axs[idx].set_title(f'm = {m/args.eV:.2f} eV, boost = {jnp.linalg.norm(boost_vec):.3f}c')
        
        axs[idx].axvline(mean_v_plot, color='black', linestyle='--', label=label)
        axs[idx].axvspan((mean_v_plot-std_v_plot), (mean_v_plot+std_v_plot), 
                        alpha=0.2, color='black', label=r'1$\sigma$ range')
        axs[idx].legend()
    
    plt.tight_layout()
    plt.show()


plot_cnb_velocity_distribution_boost(
    masses=nu_m_picks[:4], 
    unit_type="km/s",
    boost_vec=jnp.array([0,0,1])*300*Params.km/Params.s,
    args=Params())
```

```python
def plot_cnb_velocity_distributions(
        masses, 
        unit_type: str = "km/s",
        boost_vec = jnp.array([0., 0., 0.]),
        n_sigma: int = 1,
        args = None):
    
    if args is None:
        args = Params()

    v_unit = args.km/args.s
    
    def fermi_dirac(p):
        return 1 / (jnp.exp(p / (args.T_CNB)) + 1)
    
    p = jnp.geomspace(0.01, 400, 10_000)*args.T_CNB
    theta = jnp.arccos(2 * np.random.uniform(0, 1, size=len(p)) - 1)
    phi = jnp.array(np.random.uniform(0, 2*np.pi, size=len(p)))
    
    fig, axs = plt.subplots(2, 2, figsize=(12, 12))
    axs = axs.ravel()
    
    for idx, m in enumerate(masses):
        # Unboosted distribution
        f = fermi_dirac(p)
        dist = f * p**3
        E = jnp.sqrt(p**2 + m**2)
        v = p / E
        
        if unit_type == "km/s":
            v_unboosted = v/v_unit
        else:
            v_unboosted = v
            
        norm = trap(dist, jnp.log(p))
        dist_normalized = dist/norm
        
        avg_v = trap(v*dist, jnp.log(v)) / norm
        std_v = jnp.sqrt(trap((v-avg_v)**2 * dist, jnp.log(v)) / norm) * n_sigma
        
        if unit_type == "km/s":
            avg_v_plot = avg_v/v_unit
            std_v_plot = std_v/v_unit
        else:
            avg_v_plot = avg_v
            std_v_plot = std_v

        # Plot unboosted
        axs[idx].plot(v_unboosted, dist_normalized, label='Rest frame')
        axs[idx].axvline(avg_v_plot, color='blue', linestyle='--', 
                        label=f'Mean (rest): {avg_v_plot:.3f}')
        axs[idx].axvspan((avg_v_plot-std_v_plot), (avg_v_plot+std_v_plot), 
                        alpha=0.2, color='blue', label=fr'{n_sigma}$\sigma$ range (rest)')
        
        # Boosted distribution
        v_mag = p / E
        v = jnp.stack([
            v_mag * jnp.sin(theta) * jnp.cos(phi),
            v_mag * jnp.sin(theta) * jnp.sin(phi),
            v_mag * jnp.cos(theta)
        ], axis=1)
        
        v_boosted = SimUtil.S_to_Sprime_frame_trafo(v, boost_vec, args)
        v_boosted_mag = jnp.sqrt(jnp.sum(v_boosted**2, axis=1))
        sort_idx = jnp.argsort(v_boosted_mag)
        v_boosted_mag = v_boosted_mag[sort_idx]
        
        p_boosted = m * v_boosted_mag / jnp.sqrt(1 - v_boosted_mag**2)
        f = fermi_dirac(p_boosted)
        dist = f * p_boosted**3
        
        norm = trap(dist, jnp.log(p_boosted))
        dist_normalized = dist/norm
        
        mean_v = jnp.sum(v_boosted_mag * dist_normalized) / jnp.sum(dist_normalized)
        std_v = jnp.sqrt(jnp.sum((v_boosted_mag - mean_v)**2 * dist_normalized) / jnp.sum(dist_normalized)) * n_sigma
        
        if unit_type == "km/s":
            v_plot = v_boosted_mag/v_unit
            mean_v_plot = mean_v/v_unit
            std_v_plot = std_v/v_unit
        else:
            v_plot = v_boosted_mag
            mean_v_plot = mean_v
            std_v_plot = std_v
            
        # Plot boosted
        # axs[idx].plot(v_plot, dist_normalized, label='Boosted frame')
        axs[idx].axvline(mean_v_plot, color='red', linestyle='--', 
                        label=f'Mean (boost): {mean_v_plot:.3f}')
        axs[idx].axvspan((mean_v_plot-std_v_plot), (mean_v_plot+std_v_plot), 
                        alpha=0.2, color='red', label=fr'{n_sigma}$\sigma$ range (boost)')
        
        axs[idx].set_xscale('log')

        if unit_type == "km/s":
            axs[idx].set_xlabel('km/s')
            axs[idx].set_xlim(1e1, 2e5)
            boost_str = f"boost = {jnp.linalg.norm(boost_vec/(args.km/args.s)):.0f} km/s"
        else:
            axs[idx].set_xlabel('v/c')
            axs[idx].set_xlim(1e-4, 1)
            boost_str = f"boost = {jnp.linalg.norm(boost_vec):.3f}c"
            
        axs[idx].set_ylabel('Normalized Distribution')
        axs[idx].set_title(f'm = {m/args.eV:.2f} eV, {boost_str}')
        axs[idx].legend()
    
    plt.tight_layout()
    plt.show()


boost_unit = Params.km/Params.s
sun_vel_CNB = np.array([-0.06935679,-0.66222051,0.74609224])*369.82
sun_vel_MW = np.array([11.1029257,250.8323298,7.3656565])

plot_cnb_velocity_distributions(
    masses=nu_m_picks[:4], 
    unit_type="km/s",
    # unit_type="v/c",
    boost_vec=sun_vel_CNB*boost_unit,
    n_sigma=1,
    args=Params())

plot_cnb_velocity_distributions(
    masses=nu_m_picks[:4], 
    unit_type="km/s",
    # unit_type="v/c",
    boost_vec=sun_vel_MW*boost_unit,
    n_sigma=1,
    args=Params())
```

### Non-relativistic transition

```python
def get_z_nr(m_nu, p_nu):
    return m_nu/p_nu - 1

p_range = jnp.geomspace(0.01, 10, 100)*Params.T_CNB

redshifts = Utils.make_logz_arr(1e9, 0, 100)  # CNB to today
masses = nu_m_picks

# Calculate percentages
percentages = np.zeros((len(masses), len(redshifts)))
for i, m in enumerate(masses):
    z_nr = get_z_nr(m, p_range)
    for j, z in enumerate(redshifts):
        percentages[i, j] = jnp.mean(z_nr >= z) * 100

# Plot
fig = plt.figure()
fig.tight_layout()
ax = fig.add_subplot(111)
for i, m in enumerate(masses):
    ax.semilogx(1+redshifts, percentages[i, :], label=f'{m} eV')

mid_y = 10
ax.axvline(4+1, color="red", ls="dotted")
ax.text(4.9, mid_y, 'z=4', color='red', rotation=90, verticalalignment='center')
ax.axvline(31+1, color="red", ls="dotted")
ax.text(31+1e-1, mid_y, 'z=31', color='red', rotation=90, verticalalignment='center')
ax.axvline(1100+1, color="red", ls="dotted")
ax.text(1100+1e-1, mid_y, 'z=1100 (CMB)', color='red', rotation=90, verticalalignment='center')

ax.set_xlabel("1+z")
ax.set_ylabel(fr"$\%$ non-rel. $\nu$'s")
ax.invert_xaxis()
ax.legend()
ax.set_title(r"Momentum range: $0.01 \leq \frac{p_{\nu,0}}{T_{CNB,0}} \leq 10$")
plt.show(); plt.close()
```
